```python
class Conversion:
    def __init__(self, capacity):
        self.top = -1
        self.capacity = capacity
        self.array = []
        self.output = []

        self.precedence = {
            '+': 1,
            '-': 1,
            '*': 2,
            '/': 2,
            '^': 3
        }

    def isEmpty(self):
        return self.top == -1

    def peek(self):
        return self.array[-1]

    def pop(self):
        if not self.isEmpty():
            self.top -= 1
            return self.array.pop()
        return "$"

    def push(self, op):
        self.top += 1
        self.array.append(op)

    def isOperand(self, ch):
        return ch.isalnum()   # supports letters and digits

    def notGreater(self, op):
        try:
            a = self.precedence[op]
            b = self.precedence[self.peek()]
            return a <= b
        except KeyError:
            return False

    def infixToPostfix(self, exp):
        for ch in exp:
            # Operand → directly to output
            if self.isOperand(ch):
                self.output.append(ch)

            # Opening bracket
            elif ch == '(':
                self.push(ch)

            # Closing bracket
            elif ch == ')':
                while (not self.isEmpty()) and self.peek() != '(':
                    self.output.append(self.pop())
                if not self.isEmpty() and self.peek() != '(':
                    return -1
                else:
                    self.pop()  # remove '('

            # Operator
            else:
                while (not self.isEmpty()) and self.notGreater(ch):
                    self.output.append(self.pop())
                self.push(ch)

        # Pop remaining operators
        while not self.isEmpty():
            self.output.append(self.pop())

        return "".join(self.output)


# Driver code
exp = "a+b*(c^d-e)^(f+g*h)-i"
obj = Conversion(len(exp))
result = obj.infixToPostfix(exp)

print("Postfix expression:", result)
```

```python
class Evaluate:
    def __init__(self, capacity):
        self.top = -1
        self.capacity = capacity
        self.array = []

    def isEmpty(self):
        return self.top == -1

    def peek(self):
        return self.array[-1]

    def pop(self):
        if not self.isEmpty():
            self.top -= 1
            return self.array.pop()
        return "$"

    def push(self, op):
        self.top += 1
        self.array.append(op)

    def evaluatePostfix(self, exp):
        for ch in exp:
            # If operand, push to stack
            if ch.isdigit():
                self.push(int(ch))

            else:
                val1 = self.pop()
                val2 = self.pop()

                # Perform operation safely
                if ch == '+':
                    self.push(val2 + val1)
                elif ch == '-':
                    self.push(val2 - val1)
                elif ch == '*':
                    self.push(val2 * val1)
                elif ch == '/':
                    self.push(val2 // val1)  # integer division
                elif ch == '^':
                    self.push(val2 ** val1)

        return self.pop()


# Driver code
exp = "231*+9-"
obj = Evaluate(len(exp))

print("Postfix evaluation:", obj.evaluatePostfix(exp))
```
