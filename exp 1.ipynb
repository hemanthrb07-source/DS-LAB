```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at beginning
    def push(self, new_data):
        new_node = Node(new_data)
        new_node.next = self.head
        self.head = new_node

    # Insert after a given node
    def insertAfter(self, prev_node, new_data):
        if prev_node is None:
            print("The given previous node must be in LinkedList.")
            return

        new_node = Node(new_data)
        new_node.next = prev_node.next
        prev_node.next = new_node

    # Insert at end
    def append(self, new_data):
        new_node = Node(new_data)

        if self.head is None:
            self.head = new_node
            return

        last = self.head
        while last.next:
            last = last.next

        last.next = new_node

    # Print list
    def printList(self):
        temp = self.head
        while temp:
            print(temp.data)
            temp = temp.next


if __name__ == "__main__":
    my_list = LinkedList()

    my_list.append(6)
    my_list.push(7)
    my_list.push(1)
    my_list.append(4)

    # Insert 8 after second node (7)
    my_list.insertAfter(my_list.head.next, 8)

    print("Created linked list is:")
    my_list.printList()
```

```python
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None


class SLinkedList:
    def __init__(self):
        self.head = None

    def Atbegining(self, data_in):
        new_node = Node(data_in)
        new_node.next = self.head
        self.head = new_node

    def RemoveNode(self, removekey):
        head_val = self.head

        # Case 1: head node holds the key
        if head_val is not None:
            if head_val.data == removekey:
                self.head = head_val.next
                return

        # Case 2: search for the key
        prev = None
        while head_val is not None:
            if head_val.data == removekey:
                break
            prev = head_val
            head_val = head_val.next

        # If key not present
        if head_val is None:
            return

        # Unlink node
        prev.next = head_val.next

    def LListprint(self):
        printval = self.head
        while printval:
            print(printval.data)
            printval = printval.next


# Driver code
if __name__ == "__main__":
    llist = SLinkedList()

    llist.Atbegining("Mon")
    llist.Atbegining("Tue")
    llist.Atbegining("Wed")
    llist.Atbegining("Thu")

    llist.RemoveNode("Tue")

    llist.LListprint()
```
