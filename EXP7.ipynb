```python
class Node:
    # Constructor
    def __init__(self, key):
        self.data = key
        self.left = None
        self.right = None


def insert(node, data):
    if node is None:
        return Node(data)

    if data <= node.data:
        node.left = insert(node.left, data)
    else:
        node.right = insert(node.right, data)

    return node


def minValue(node):
    current = node

    while current.left is not None:
        current = current.left

    return current.data


# Driver Code
root = None
root = insert(root, 4)
insert(root, 2)
insert(root, 1)
insert(root, 3)
insert(root, 6)
insert(root, 5)

print("Minimum value in BST is %d" % (minValue(root)))
```

```python
INT_MIN = float('-inf')

def canRepresentBST(pre):
    s = []
    root = INT_MIN

    for value in pre:
        # If we find a node who is on the right side
        # and smaller than root, return False
        if value < root:
            return False

        # Nodes popped are ancestors of current node
        while len(s) > 0 and s[-1] < value:
            root = s.pop()

        # Push current value to stack
        s.append(value)

    return True


# Test Cases
pre1 = [40, 30, 35, 80, 100]
print("true" if canRepresentBST(pre1) else "false")

pre2 = [40, 30, 35, 20, 80, 100]
print("true" if canRepresentBST(pre2) else "false")
```

```python
class newNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def getCount(root, low, high):
    if root is None:
        return 0

    # If node is within range
    if low <= root.data <= high:
        return (1 +
                getCount(root.left, low, high) +
                getCount(root.right, low, high))

    # If node is smaller than range → ignore left subtree
    elif root.data < low:
        return getCount(root.right, low, high)

    # If node is greater than range → ignore right subtree
    else:
        return getCount(root.left, low, high)


# Driver Code
if __name__ == "__main__":
    root = newNode(10)
    root.left = newNode(5)
    root.right = newNode(50)
    root.left.left = newNode(1)
    root.right.left = newNode(40)
    root.right.right = newNode(100)

    l = 5
    h = 45

    print("Count of nodes between [", l, ",", h, "] is",
          getCount(root, l, h))
```
