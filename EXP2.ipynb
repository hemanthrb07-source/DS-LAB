```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        new_node = Node(data)


        if self.head is None:
            self.head = new_node
            self.tail = new_node
            new_node.next = self.head
        else:

            self.tail.next = new_node
            self.tail = new_node
            self.tail.next = self.head

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        print("Nodes of the circular linked list:")

        while True:
            print(current.data)
            current = current.next
            if current == self.head:
                break


if __name__ == "__main__":
    cl = CircularLinkedList()

    cl.add(1)
    cl.add(2)
    cl.add(3)
    cl.add(4)

    cl.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.last = None


    def addToEmpty(self, data):
        if self.last is not None:
            return self.last

        temp = Node(data)
        self.last = temp
        self.last.next = self.last
        return self.last


    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        self.last = temp
        return self.last


    def addAfter(self, data, item):
        if self.last is None:
            return None

        p = self.last.next

        while True:
            if p.data == item:
                temp = Node(data)
                temp.next = p.next
                p.next = temp

                if p == self.last:
                    self.last = temp
                return self.last

            p = p.next
            if p == self.last.next:
                break

        return None

    def display(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next
        while True:
            print(temp.data)
            temp = temp.next
            if temp == self.last.next:
                break


if __name__ == "__main__":
    cl = CircularLinkedList()

    cl.addToEmpty(1)
    cl.addEnd(2)
    cl.addEnd(3)
    cl.addBegin(0)
    cl.addAfter(5, 2)

    cl.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None


    def add(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
            new_node.next = self.head
        else:
            self.tail.next = new_node
            self.tail = new_node
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head is None:
            return


        if self.head == self.tail:
            self.head = None
            self.tail = None
            return


        current = self.head
        while current.next != self.tail:
            current = current.next

        self.tail = current
        self.tail.next = self.head

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        while True:
            print(current.data)
            current = current.next
            if current == self.head:
                break
        print()


if __name__ == "__main__":
    cl = CircularLinkedList()

    cl.add(1)
    cl.add(2)
    cl.add(3)
    cl.add(4)

    print("Original List:")
    cl.display()

    # Repeatedly delete last node
    while cl.head is not None:
        cl.deleteEnd()
        print("Updated List:")
        cl.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def push(self, data):
        ptr1 = Node(data)

        if self.head is None:
            ptr1.next = ptr1
            self.head = ptr1
            return

        temp = self.head
        while temp.next != self.head:
            temp = temp.next

        ptr1.next = self.head
        temp.next = ptr1
        self.head = ptr1

    def printList(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while True:
            print(temp.data)
            temp = temp.next
            if temp == self.head:
                break


if __name__ == "__main__":
    cllist = CircularLinkedList()

    cllist.push(12)
    cllist.push(56)
    cllist.push(2)
    cllist.push(11)

    print("Contents of circular linked list:")
    cllist.printList()
```
